## 1 · Ponto de partida e estratégia

O guião é explícito: «*Starting from the Wikipedia Main Page*» e «*Locations and coordinates must be
obtained only through this method*».

Documentar aqui, **em texto**, o caminho que vai ser seguido da página principal até à página de um
artista, e depois até às coordenadas do local. Esta explicação faz parte da avaliação do tópico
avançado — a grelha quer ver que perceberam o mecanismo, e não que reutilizaram código.

**Resultado esperado:** a estratégia escrita e justificada, antes de existir uma linha de código.

## 2 · Chegar à página de cada artista

### 2.1 Localizar um artista (caso único)
Fazer funcionar para **um** artista antes de pensar em repetir. É aqui que se descobrem os problemas.

### 2.2 Generalizar para a lista completa
Repetir para todos os artistas, guardando também os casos em que não se encontrou página.

**Resultado esperado:** uma correspondência artista → página, e a lista dos que ficaram por resolver.

## 3 · Extrair o local de nascimento ou origem

### 3.1 Identificar onde está a informação na página
Pode estar em sítios diferentes consoante o artista seja uma pessoa ou um grupo — para um grupo não
há local de nascimento. Decidir o que se usa em cada caso e **escrever a decisão**.

### 3.2 Extrair e normalizar
O mesmo local pode aparecer escrito de várias maneiras. Normalizar antes de procurar coordenadas,
para não procurar o mesmo sítio várias vezes.

**Resultado esperado:** uma coluna de local por artista, e a contagem dos que não foi possível obter.

## 4 · Validar a correspondência artista → local

O guião faz disto um requisito à parte: «*Ensure that artists are correctly matched to their
birthplace or origin*». Verificar uma amostra manualmente e descrever o método de verificação.

**Resultado esperado:** evidência de que a correspondência está certa, não só a afirmação de que está.

## 5 · Coordenadas dos locais

### 5.1 Extrair as coordenadas
Pelo mesmo método, a partir da Wikipédia. Procurar por **local distinto** e não por artista, para
não repetir trabalho.

### 5.2 Validar a correspondência local → coordenadas
Também é um requisito explícito. Confirmar uma amostra e descrever como.

**Resultado esperado:** uma tabela de locais com latitude e longitude, verificada.

## 6 · Cobertura

O guião reconhece que não será possível resolver todos os casos, mas diz que «*exploring how to
reach as many as possible within Wikipedia will be rewarded*».

Quantificar: quantos artistas ficaram com local, quantos locais ficaram com coordenadas, e **porque
falharam os restantes**. Se foram tentadas abordagens alternativas para recuperar casos, descrevê-las
aqui — é exatamente o que a grelha premeia.

**Resultado esperado:** números de cobertura e uma explicação das falhas.
