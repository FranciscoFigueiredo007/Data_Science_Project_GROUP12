## 0 · Setup

Importar as bibliotecas, mostrar a versão de cada pacote usado (requisito do guião) e ler os dados
de entrada.

## 1 · Juntar as coordenadas aos artistas

Juntar as duas tabelas e verificar quantos artistas ficam efetivamente com coordenada.

Decidir — e escrever — o que fazer aos artistas sem coordenada: ficam de fora dos mapas, mas devem
continuar a contar nas análises? A decisão tem de ser coerente com o resto do trabalho.

**Resultado esperado:** uma tabela pronta a mapear, e o número de artistas que ficaram de fora e porquê.

## 2 · Preparar os campos do hover

O guião exige quatro informações em cada marcador: nome do artista, local de nascimento ou origem,
total de streams, e a **percentagem de streams a solo e em colaboração**.

As percentagens não existem no dataset — têm de ser calculadas aqui, a partir das colunas de streams.

**Resultado esperado:** uma coluna por cada um dos quatro campos, com os números já formatados como
vão aparecer no mapa.

## 3 · Mapa 1 — marcadores por artista

### 3.1 Desenhar o mapa base
Um marcador por artista, posicionado pelas coordenadas.

### 3.2 Dimensionar os marcadores
O tamanho tem de depender do **total de streams**. Pensar na escala: com valores muito diferentes,
os marcadores pequenos desaparecem e os grandes tapam tudo.

### 3.3 Hover ou clique
Ligar os quatro campos da secção 2.

**Resultado esperado:** um mapa navegável em que, ao passar o rato num artista, aparecem as quatro
informações pedidas.

## 4 · Mapa 2 — densidade

A distribuição global dos streams com base no local de nascimento ou origem dos artistas.

Decidir o que é a densidade neste caso — número de artistas, ou streams acumulados? O guião fala em
«*global distribution of streams*», o que aponta para a segunda hipótese. Escrever a decisão.

**Resultado esperado:** um mapa de densidade legível, com a escolha justificada.

## 5 · Além do mínimo

information is provided*», e a grelha premeia originalidade e melhorias acima do baseline.

Espaço para o que o grupo quiser acrescentar — filtros, cores por género ou por língua, um terceiro
mapa. **Só depois de os dois mapas obrigatórios estarem completos.**
